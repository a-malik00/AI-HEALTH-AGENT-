```python
from pydantic import BaseModel, Field
from typing import Optional, List
from enum import Enum

class TrustLevel(str, Enum):
    HIGH = "HIGH"
    MEDIUM = "MEDIUM"
    LOW = "LOW"
    SUSPICIOUS = "SUSPICIOUS"

class FacilityRecord(BaseModel):
    # Basic info
    facility_id: str
    name: str
    district: str
    state: str
    pin_code: Optional[str] = None
    facility_type: Optional[str] = None  # PHC, CHC, District Hospital, etc.

    # Capabilities extracted from notes
    has_icu: bool = False
    has_ot: bool = False             # Operation Theatre
    has_nicu: bool = False
    has_dialysis: bool = False
    has_oncology: bool = False
    has_blood_bank: bool = False
    has_oxygen_supply: bool = False
    is_24_7: bool = False

    # Staff
    has_anesthesiologist: bool = False
    has_surgeon: bool = False
    doctor_count: Optional[int] = None
    uses_parttime_doctors: bool = False

    # Equipment (raw list from notes)
    equipment_mentioned: List[str] = Field(default_factory=list)
    specialties_mentioned: List[str] = Field(default_factory=list)

    # Trust scoring (filled by Validator Agent)
    trust_score: Optional[float] = None
    trust_level: Optional[TrustLevel] = None
    contradictions: List[str] = Field(default_factory=list)
    trust_reasoning: Optional[str] = None

    # Original text for traceability
    raw_notes: Optional[str] = None
    data_staleness_years: Optional[float] = None  # temporal decay flag

print("Schema defined. Fields:", list(FacilityRecord.model_fields.keys()))
```
